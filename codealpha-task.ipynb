```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df["resting_bp"] = scaler.fit_transform(df[["resting_bp"]])
print(df["resting_bp"])
df["oldpeak"] = scaler.fit_transform(df[["oldpeak"]])
df["cholestoral"] = scaler.fit_transform(df[["cholestoral"]])
```

```python
X = df[["age","sex","chest_pain_type","resting_bp","cholestoral","fasting_blood_sugar","restecg","max_hr","oldpeak","num_major_vessels"]]
y = df["target"]
```

```python
plt.scatter(df["age"],y,color="pink",alpha=0.8)
plt.title("AGE VS CHOLESTROL")
plt.xlabel("AGE")
plt.ylabel("CHOLESTROL")
plt.show()
```

```python
columns = ["age","resting_bp","cholestoral","max_hr","oldpeak"]
for col in columns:
    df[col].plot(kind="box")
    plt.title(col)
    plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size =0.3,random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X_train,y_train)
lr_pred = lr.predict(X_test)
print(lr_pred)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier(n_estimators=100,random_state=42)
rfc.fit(X_train,y_train)
rfc_pred = rfc.predict(X_test)
print(rfc_pred)
```

```python
from sklearn.svm import SVC
svc = SVC(kernel="rbf",C=100,random_state=42,probability=True)
svc.fit(X_train,y_train)
svc_pred = svc.predict(X_test)
print(svc_pred)
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier(
    n_estimators=100,      # Number of trees
    learning_rate=0.05,    # Step size shrinkage to prevent overfitting
    max_depth=3,           # Depth of each tree
    random_state=42
)
xgb.fit(X_train,y_train)
xgb_pred = xgb.predict(X_test)
print(xgb_pred)
```

```python
from sklearn.metrics import accuracy_score,recall_score,precision_score,classification_report,confusion_matrix
print("EVALUATION OF LOGISTICREGRESSION")
accuracy_lr = accuracy_score(y_test,lr_pred)*100
print("ACCURACY IS=",accuracy_lr)
print("PRECISION IS=",precision_score(y_test,lr_pred)*100)
print("RECALL IS=",recall_score(y_test,lr_pred)*100)
print("CLASSIFICATION REPORT",classification_report(y_test,lr_pred))
print("CONFUSION MATRIX",confusion_matrix(y_test,lr_pred))

print("EVALUATION OF RANDOMFORESTCLASSIFIER")
accuracy_rfc = accuracy_score(y_test,rfc_pred)*100
print("ACCURACY IS=",accuracy_rfc)
print("PRECISION IS=",precision_score(y_test,rfc_pred)*100)
print("RECALL IS=",recall_score(y_test,rfc_pred)*100)
print("CLASSIFICATION REPORT",classification_report(y_test,rfc_pred))
print("CONFUSION MATRIX",confusion_matrix(y_test,rfc_pred))


print("EVALUATION OF SUPPORTYVECTORCLASSIFIER")
accuracy_svm = accuracy_score(y_test,svc_pred)*100
print("ACCURACY IS=",accuracy_svm)
print("PRECISION IS=",precision_score(y_test,svc_pred)*100)
print("RECALL IS=",recall_score(y_test,svc_pred)*100)
print("CLASSIFICATION REPORT",classification_report(y_test,svc_pred))
print("CONFUSION MATRIX",confusion_matrix(y_test,svc_pred))


print("EVALUATION OF XGBCLASSIFIER")
accuracy_xgb = accuracy_score(y_test,xgb_pred)*100
print("ACCURACY IS=",accuracy_xgb)
print("PRECISION IS=",precision_score(y_test,xgb_pred)*100)
print("RECALL IS=",recall_score(y_test,xgb_pred)*100)
print("CLASSIFICATION REPORT",classification_report(y_test,xgb_pred))
print("CONFUSION MATRIX",confusion_matrix(y_test,xgb_pred))
```

```python
train_accuracy_rfc = rfc.score(X_train,y_train)
test_accuracy_rfc = rfc.score(X_test,y_test)
print(train_accuracy_rfc*100)
print(test_accuracy_rfc*100)

train_accuracy_lr = lr.score(X_train,y_train)
test_accuracy_lr = lr.score(X_test,y_test)
print(train_accuracy_lr*100)
print(test_accuracy_lr*100)

train_accuracy_svc = svc.score(X_train,y_train)
test_accuracy_svc = svc.score(X_test,y_test)
print(train_accuracy_svc*100)
print(test_accuracy_svc*100)

train_accuracy_xgb = xgb.score(X_train,y_train)
test_accuracy_xgb = xgb.score(X_test,y_test)
print(train_accuracy_xgb*100)
print(test_accuracy_xgb*100)
```

```python
from sklearn.model_selection import cross_val_score
models = {
    "Logistic Regression": LogisticRegression(),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "SVM": SVC(kernel="rbf", C=1.0, random_state=42,probability=True),
    "XGBoost": XGBClassifier(n_estimators=100, learning_rate=0.05, max_depth=3, random_state=42)
}
for name,model in models.items():
  scores = cross_val_score(model,X_train,y_train,cv=5,scoring="accuracy")
  print(name)
  print(scores)
  plt.plot([1,2,3,4,5],scores,marker="o")
plt.xlabel("Fold")
plt.ylabel("Accuracy")
plt.title("Cross Validation Scores")
plt.show()
```

```python
from sklearn.metrics import roc_auc_score, roc_curve
import matplotlib.pyplot as plt

# List of models aur unke predictions
models_list = [("Logistic", lr), ("RandomForest", rfc), ("SVM", svc), ("XGBoost", xgb)]

plt.figure(figsize=(10,8))

for name, model in models_list:
    # 1. Probabilities nikalo
    y_prob = model.predict_proba(X_test)[:,1]
    
    # 2. ROC curve ke points nikalo
    fpr, tpr, _ = roc_curve(y_test, y_prob)
    
    # 3. AUC score nikalo
    auc_val = roc_auc_score(y_test, y_prob)
    
    # 4. Plot karo
    plt.plot(fpr, tpr, label=f"{name} AUC = {auc_val:.3f}")

# Random line
plt.plot([0,1], [0,1], 'k--', lw=2) 
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - All Models")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```
